```python
import pyslim
import tskit
import msprime
import numpy as np

scenarios = ["scenario1", "scenario2", "scenario3"]

for scenario in scenarios:
    print(f"--- Processing {scenario} ---")
    
    # 1. Load the tree sequence for the current scenario
    input_file = f"output.{scenario}.trees"
    output_file = f"output.{scenario}.vcf"
    
    try:
        ts = tskit.load(input_file)
    except FileNotFoundError:
        print(f"File {input_file} not found. Skipping...")
        continue

    # 2. Recapitated
    ts = pyslim.recapitate(ts, recombination_rate=1e-8, ancestral_Ne=10000)

    # 3. Select 10 individuals per population (IDs 1, 2, 3)
    selected_individuals = []
    for pop_id in [1, 2, 3]:
        pop_inds = [ind for ind in ts.individuals() if ind.population == pop_id]
        
        if len(pop_inds) < 10:
            print(f"Warning: Only found {len(pop_inds)} individuals in pop {pop_id}")
        
        selected_individuals.extend(pop_inds[:10])

    # 4. Get the NODE IDs for those individuals
    node_ids = []
    for ind in selected_individuals:
        node_ids.extend(ind.nodes)

    # 5. Simplify the tree
    ts = ts.simplify(node_ids)

    # 6. Overlay mutations
    ts = msprime.mutate(ts, rate=1e-7, random_seed=42)

    # 7. Generate the 30 names
    sample_names = []
    for p_num in [1, 2, 3]:
        for i in range(10):
            sample_names.append(f"p{p_num}_{i}")

    # 8. Final VCF Output
    if ts.num_individuals == len(sample_names):
        with open(output_file, "w") as vcf_file:
            ts.write_vcf(vcf_file, individual_names=sample_names)
        print(f"Success! {output_file} created with {ts.num_mutations} mutations.")
    else:
        print(f"Error: Individual count ({ts.num_individuals}) doesn't match name count ({len(sample_names)}) for {scenario}.")

print("\nAll scenarios processed.")
```
